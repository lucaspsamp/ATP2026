### TPC5: Aplicação para gestão de alunos

Considere que o modelo do aluno e da turma têm a seguinte estrutura:

`aluno = (nome, id, [notaTPC, notaProj, notaTeste])`

`turma = [aluno]`

* Cria uma aplicação que coloca no monitor o seguinte menu de operações:
    - 1: Criar uma turma;
    - 2: Inserir um aluno na turma;
    - 3: Listar a turma;
    - 4: Consultar um aluno por id;
    - 8: Guardar a turma em ficheiro;
    - 9: Carregar uma turma dum ficheiro;
    - 0: Sair da aplicação
* No fim de executar a operação selecionada, a aplicação deverá colocar novamente o menu e pedir ao utilizador a opção para continuar;
* Utiliza a tua aplicação para criar uma turma com 5 alunos.

```python
turma = []

opcao = -1

while opcao != 0:

    print("\n----- MENU -----")
    print("1 - Criar turma")
    print("2 - Inserir aluno")
    print("3 - Listar turma")
    print("4 - Consultar aluno por ID")
    print("8 - Guardar turma em ficheiro")
    print("9 - Carregar turma de ficheiro")
    print("0 - Sair")

    opcao = int(input("Opção: "))

    if opcao == 1:

        turma = []

        print("Turma criada com sucesso")

    elif opcao == 2:

        nome = input("Nome: ")
        numero = int(input("ID: "))

        notaTPC = int(input("Nota TPC: "))
        notaProj = int(input("Nota Projeto: "))
        notaTeste = int(input("Nota Teste: "))

        aluno = (nome, numero, [notaTPC, notaProj, notaTeste])

        turma.append(aluno)

        print("Aluno inserido")

    elif opcao == 3:

        for aluno in turma:

            print("Nome:", aluno[0])
            print("ID:", aluno[1])
            print("Notas:", aluno[2])
            print()

    elif opcao == 4:

        procura = int(input("ID a procurar: "))

        encontrado = False

        for aluno in turma:

            if aluno[1] == procura:

                print("Nome:", aluno[0])
                print("ID:", aluno[1])
                print("Notas:", aluno[2])

                encontrado = True

        if encontrado == False:
            print("Aluno não encontrado")

    elif opcao == 8:

        ficheiro = open("turma.txt", "w")

        for aluno in turma:

            ficheiro.write(str(aluno))
            ficheiro.write("\n")

        ficheiro.close()

        print("Turma guardada")

    elif opcao == 9:

        ficheiro = open("turma.txt", "r")

        linhas = ficheiro.readlines()

        turma = []

        for linha in linhas:

            aluno = eval(linha.strip())
            turma.append(aluno)

        ficheiro.close()

        print("Turma carregada")

    elif opcao == 0:

        print("Programa terminado")

    else:

        print("Opção inválida")
```
