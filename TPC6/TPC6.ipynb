### TPC6: Resolver a alínea anterior

Cria uma aplicação Python que permita ao utilizador usar todas as funcionalidades pedidas nesta ficha.


```python
from matplotlib import pyplot as plt

def medias(tabMeteo):

    res = []

    for dia in tabMeteo:

        tmin = dia[1]
        tmax = dia[2]

        media = (tmin + tmax) / 2

        tuplo = (dia[0], media)

        res.append(tuplo)

    return res

def guardaTabMeteo(t, fnome):

    file = open(fnome, "w")

    for data, tmin, tmax, prec in t:

        ano, mes, dia = data

        file.write(f"{ano}-{mes}-{dia}|{tmin}|{tmax}|{prec}\n")

    file.close()

    return


def carregaTabMeteo(fnome):

    res = []

    file = open(fnome, "r")

    for line in file:

        data, tmin, tmax, prec = line.split("|")

        ano, mes, dia = data.split("-")

        tuplo = (
            (int(ano), int(mes), int(dia)),
            float(tmin),
            float(tmax),
            float(prec)
        )

        res.append(tuplo)

    file.close()

    return res


def minMin(tabMeteo):

    minima = tabMeteo[0][1]

    for dia in tabMeteo:

        if dia[1] < minima:
            minima = dia[1]

    return minima


def amplTerm(tabMeteo):

    res = []

    for dia in tabMeteo:

        data = dia[0]
        tmin = dia[1]
        tmax = dia[2]

        amplitude = tmax - tmin

        tuplo = (data, amplitude)

        res.append(tuplo)

    return res


def maxChuva(tabMeteo):

    max_prec = tabMeteo[0][3]
    max_data = tabMeteo[0][0]

    for dia in tabMeteo:

        if dia[3] > max_prec:

            max_prec = dia[3]
            max_data = dia[0]

    return (max_data, max_prec)

def diasChuvosos(tabMeteo, p):

    res = []

    for dia in tabMeteo:

        data = dia[0]
        prec = dia[3]

        if prec > p:

            tuplo = (data, prec)

            res.append(tuplo)

    return res


def maxPeriodoCalor(tabMeteo, p):

    atual = 0
    maior = 0

    for dia in tabMeteo:

        prec = dia[3]

        if prec < p:

            atual = atual + 1

            if atual > maior:
                maior = atual

        else:
            atual = 0

    return maior


def grafTabMeteo(t):

    x = []
    y_min = []
    y_max = []
    y_prec = []

    for data, tmin, tmax, prec in t:

        ano, mes, dia = data

        x.append(f"{dia}/{mes}")

        y_min.append(tmin)
        y_max.append(tmax)
        y_prec.append(prec)

    plt.plot(x, y_min, marker="o", label="Tmin")
    plt.plot(x, y_max, marker="o", label="Tmax")

    plt.legend()
    plt.grid()

    plt.show()

    plt.bar(x, y_prec)

    plt.title("Precipitacao")
    plt.grid()

    plt.show()

    return

def criaTabMeteo():

    tab = []

    n = int(input("Quantos dias pretende inserir? "))

    for i in range(n):

        ano = int(input("Ano: "))
        mes = int(input("Mes: "))
        dia = int(input("Dia: "))

        tmin = float(input("Temp minima: "))
        tmax = float(input("Temp maxima: "))
        prec = float(input("Precipitacao: "))

        registo = ((ano, mes, dia), tmin, tmax, prec)

        tab.append(registo)

    return tab


tabMeteo = []

opcao = -1

while opcao != 0:

    print()
    print("===== MENU =====")
    print("1 - Criar tabela")
    print("2 - Medias")
    print("3 - Guardar tabela")
    print("4 - Carregar tabela")
    print("5 - Minima mais baixa")
    print("6 - Amplitude termica")
    print("7 - Maxima precipitacao")
    print("8 - Dias chuvosos")
    print("9 - Maior periodo seco")
    print("10 - Graficos")
    print("0 - Sair")

    opcao = int(input("Opcao: "))

    if opcao == 1:

        tabMeteo = criaTabMeteo()

    elif opcao == 2:

        print(medias(tabMeteo))

    elif opcao == 3:

        nome = input("Nome do ficheiro: ")

        guardaTabMeteo(tabMeteo, nome)

    elif opcao == 4:

        nome = input("Nome do ficheiro: ")

        tabMeteo = carregaTabMeteo(nome)

        print(tabMeteo)

    elif opcao == 5:

        print(minMin(tabMeteo))

    elif opcao == 6:

        print(amplTerm(tabMeteo))

    elif opcao == 7:

        print(maxChuva(tabMeteo))

    elif opcao == 8:

        p = float(input("Limite: "))

        print(diasChuvosos(tabMeteo, p))

    elif opcao == 9:

        p = float(input("Limite: "))

        print(maxPeriodoCalor(tabMeteo, p))

    elif opcao == 10:

        grafTabMeteo(tabMeteo)

    elif opcao == 0:

        print("Fim.")

    else:

        print("Opcao invalida.")
```
